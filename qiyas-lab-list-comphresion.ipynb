```python
#traditional list loop way
num = [1,2,3,4,5]
squares = []
for n in num:
    squares.append(n ** 2)
print(squares)
```

```python
#list comphresion loop way

num = [1,2,3,4,5]
squares = [n ** 2 for n in num]
print(squares)
```

```python
#flattening a matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

flat = [num for row in matrix for num in row]

print(flat)
```

```python
#conditional extraction from dictionary
scores = {
    "Alice": 90,
    "Bob": 65,
    "Charlie": 85
}

passed = [name for name, score in scores.items() if score >= 70]
passed_indexed = [f"{name}: {score}" for name,score in enumerate(scores)]
print(passed_indexed)

print(passed)
```

```python
#list comphresion with enumertae methos
names = ["Alice", "Bob", "Charlie"]

indexed = [
    f"{i}: {name}"
    for i, name in enumerate(names)
]

print(indexed)
```
