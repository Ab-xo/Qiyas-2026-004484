```python
import requests
from bs4 import BeautifulSoup
import csv

url = "https://www.worldometers.info/coronavirus/"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table", id="main_table_countries_today")

# Open CSV file
with open("covid_data.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    # Header row
    writer.writerow([
        "Country",
        "Total Cases",
        "New Cases",
        "Total Deaths",
        "New Deaths",
        "Total Recovered",
        "Active Cases",
        "Critical Cases"
    ])

    # Extract rows
    rows = table.tbody.find_all("tr")
    data_rows = []

    for row in rows:
        cols = row.find_all("td")
        if len(cols) > 10:
            data_rows.append({
                "country": cols[1].text.strip(),
                "cols": cols
            })

    # Find USA start index
    start_index = 0
    end_index = len(data_rows)
    for i, row in enumerate(data_rows):
        if row["country"] == "USA":
            start_index = i
            break
            
        if row["country"] == "MS Zaandam":
            end_index = i
            break

    # Reorder (USA first)
    ordered_rows = data_rows[start_index:end_index +1 ]

    # Write data
    for row in ordered_rows:
        cols = row["cols"]

        writer.writerow([
            cols[1].text.strip(),
            cols[2].text.strip(),
            cols[3].text.strip(),
            cols[4].text.strip(),
            cols[5].text.strip(),
            cols[6].text.strip(),
            cols[8].text.strip(),
            cols[9].text.strip()
        ])

print("Done")
```
