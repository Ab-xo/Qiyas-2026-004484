# University Student Data — Cleaning & Analysis

This notebook builds a small student dataset, cleans the missing values,
and then runs a few summary analyses on it.

**Note on running this notebook:** the cells below are split one-task-per-cell
(matching the assignment's numbered steps) on purpose, so each transformation
is visible separately. They must be run **in order, top to bottom** (or via
*Run All*), because each cleaning step modifies `university_df` in place and
later cells depend on the result of earlier ones.


## Setup: create the dataset


```python
# Import required libraries
import pandas as pd
import numpy as np

# Create the university student dataset
university_df = pd.DataFrame({
    "student_id": [201, 202, 203, 204, 205, 206, 207, 208, 209, 210],
    "name": [
        "Abel", "Sara", None, "John", "Marta",
        None, "David", "Helen", "Tom", None
    ],
    "department": [
        "CS", "IT", "CS", None, "SE",
        "IT", "SE", None, "CS", "IT"
    ],
    "gpa": [
        3.5, 3.8, np.nan, 2.9, 3.7,
        np.nan, 3.2, 3.4, 3.9, np.nan
    ],
    "scholarship": [
        5000, np.nan, 3000, 2000, np.nan,
        4000, 3500, np.nan, 5000, np.nan
    ]
})

university_df
```

```python
# Check dataset structure and information
university_df.info()
```

## Data Cleaning


```python
# 1. Count missing values in each column
missing_count = university_df.isnull().sum()
print("Missing values per column:")
missing_count
```

```python
# 2. Calculate missing value percentage for each column
missing_percentage = (missing_count / len(university_df)) * 100
print("Missing value percentage (%) per column:")
missing_percentage
```

```python
# 3. Replace missing names with "Unknown"
university_df["name"] = university_df["name"].fillna("Unknown")
university_df
```

```python
# 4. Replace missing departments using the mode (most common department)
university_df["department"] = university_df["department"].fillna(
    university_df["department"].mode()[0]
)
university_df
```

```python
# 5. Replace missing GPA values using the median GPA
university_df["gpa"] = university_df["gpa"].fillna(
    university_df["gpa"].median()
)
university_df
```

```python
# 6. Replace missing scholarship values with zero
university_df["scholarship"] = university_df["scholarship"].fillna(0)
university_df
```

## Data Analysis


```python
# 7. Count students by department
department_counts = university_df["department"].value_counts()
print("Students by department:")
print(department_counts)
```

```python
# 8. Count scholarship recipients
num_scholarship_recipients = (university_df["scholarship"] > 0).sum()
print("Number of scholarship recipients:", num_scholarship_recipients)
```

```python
# 9. Find the highest GPA
highest_gpa = university_df["gpa"].max()
print("Highest GPA:", highest_gpa)
```

```python
# 10. Find the lowest GPA
lowest_gpa = university_df["gpa"].min()
print("Lowest GPA:", lowest_gpa)
```

```python
# 11. Calculate the average GPA
average_gpa = university_df["gpa"].mean()
print("Average GPA:", round(average_gpa, 2))
```

```python
# 12. Calculate the total scholarship amount awarded
total_scholarship = university_df["scholarship"].sum()
print("Total scholarship amount:", total_scholarship)
```

## Final Cleaned Dataset


```python
# Final check: the fully cleaned dataset
print("Final cleaned dataset:")
university_df
```
