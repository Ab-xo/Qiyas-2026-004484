### Exercise 23: Cleaning a Customer Dataset

### Exercise 24: Outlier Detection Using IQR
**Task:** Use the interquartile range (IQR) method to flag outliers in a sales series.

```python
sales = pd.Series([100, 120, 110, 105, 130, 115, 108, 500])

q1 = sales.quantile(0.25)
q3 = sales.quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = sales[(sales < lower_bound) | (sales > upper_bound)]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("\nOutliers:")
print(outliers)
```

### Exercise 25: Outlier Handling & Complete Cleaning Pipeline
**Task:** Combine every technique from this section into a single, reusable cleaning
pipeline: handle missing values, standardize text, remove duplicates, and remove
outliers.

To keep the logic easy to read, test, and reuse, the pipeline is written as a function
rather than a long sequence of standalone steps.

```python
def clean_dataset(df: pd.DataFrame) -> pd.DataFrame:
    """
    Apply a standard cleaning pipeline to a messy dataset:

      1. Impute missing numeric values (Age) with the column median.
      2. Impute missing text values (Name) with a placeholder.
      3. Standardize text casing.
      4. Remove duplicate rows.
      5. Remove outliers in the Sales column using the IQR method.

    Parameters
    ----------
    df : pd.DataFrame
        Must contain "Name", "Age", and "Sales" columns.

    Returns
    -------
    pd.DataFrame
        A cleaned copy of the input data, with a fresh index.
    """
    cleaned = df.copy()

    # 1-2: Handle missing values
    cleaned["Age"] = cleaned["Age"].fillna(cleaned["Age"].median())
    cleaned["Name"] = cleaned["Name"].fillna("Unknown")

    # 3: Standardize text
    cleaned["Name"] = cleaned["Name"].str.title()

    # 4: Remove duplicates
    cleaned = cleaned.drop_duplicates()

    # 5: Remove outliers (IQR method) on Sales
    q1, q3 = cleaned["Sales"].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower_bound, upper_bound = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    cleaned = cleaned[cleaned["Sales"].between(lower_bound, upper_bound)]

    return cleaned.reset_index(drop=True)


# Messy sample dataset: missing values, duplicates, and a Sales outlier
messy_data = pd.DataFrame({
    "Name": ["john", "John", "sara", "sara", None],
    "Age": [25, 25, np.nan, np.nan, 300],
    "Sales": [100, 100, 120, 120, 5000],
})

print("Original dataset:")
print(messy_data)

print("\nMissing values per column:")
print(messy_data.isnull().sum())

final_data = clean_dataset(messy_data)

print("\nFinal cleaned dataset:")
print(final_data)
```
